## Part 6: Linear LSE
### (35 marks)

### Curve Fitting

curvefitting.png

### Observations: values of a function f(a) in range of -π to π, contaminated with random noise

### Unknown States: coefficients of polynomial functions for curve fitting

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
df_data = df_data = pd.read_excel(/content/ENGG680_2025_Fall/Assignment 1/data.xlsx)
arr_data = df_data.to_numpy()

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
l = arr_data[:, 0]
l_var = arr_data[:, 1]
l_true = arr_data[:, 2]
```

```python
# step 3
# define the number of observation n
# (1 mark)
n = len(l)   # the number of observations equals the length of your noisy observation array

# define the observation weight matrix P for LSE as identity matrix
# identity weight matrix simply means the observations will be unweighted in the following LSE
# (1 mark)
P = np.eye(n)   # creates an n x n identity matrix

# define the range of function input(i.e., constants a), from start to end; and the numerical array of a
start = -1 * np.pi
end = np.pi
a = np.linspace(start, end, n)
```

```python
import matplotlib.pyplot as plt

# Step 4
# use matplotlib to plot the true and noisy observations together in one single figure
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# Plot the true observations
ax.plot(a, l_true, label="f(a)", color='k')  # 'k' = black

# Plot the noisy observations
ax.scatter(a, l, label="f(a) + e", color='tab:blue', marker='o', alpha=0.3)

# Add labels, title, and legend
ax.set_xlabel("a (radians)")
ax.set_ylabel("Observation values")
ax.set_title("True vs Noisy Observations")
ax.legend()

# Set x-axis range from -1 to 1
ax.set_xlim(-1, 1)

# Add grid and layout
ax.grid(True, which='both', ls='-', color='0.65')
fig.tight_layout()

plt.show()
```

```python
def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients (covariance matrix)
        apv:            a posteriori variance factor (goodness of fit)
        v:              observation residuals
    '''

    # 1️⃣ Build the design matrix A for a polynomial of degree `deg`
    #    Each column is a^0, a^1, ..., a^deg
    A = np.vander(a, deg + 1, increasing=True)

    # 2️⃣ Compute the Normal Equation components
    #    (Aᵀ * P * A) * x = Aᵀ * P * l
    N = A.T @ P @ A           # Normal matrix
    nL = A.T @ P @ l          # Right-hand side of normal equations

    # 3️⃣ Solve for polynomial coefficients x (least-squares solution)
    x = np.linalg.inv(N) @ nL

    # 4️⃣ Compute residuals (v = A*x - l)
    v = A @ x - l

    # 5️⃣ Compute a posteriori variance factor (goodness of fit)
    #    apv = (vᵀ * P * v) / redundancy, where redundancy = n - (deg + 1)
    n = len(l)
    redundancy = n - (deg + 1)
    apv = (v.T @ P @ v) / redundancy

    # 6️⃣ Compute covariance matrix of estimated coefficients
    cov_x = apv * np.linalg.inv(N)

    # Return all required outputs
    return (x, cov_x, apv, v)
```

```python
# Step 6
# Call the fit_polynomial_curve function and fetch the results
# Use polynomial degrees of: 2, 4, 6
# (1 mark)

# Fit polynomial of degree 2
x_deg2, cov_deg2, apv_deg2, v_deg2 = fit_polynomial_curve(a, l, P, deg=2)

# Fit polynomial of degree 4
x_deg4, cov_deg4, apv_deg4, v_deg4 = fit_polynomial_curve(a, l, P, deg=4)

# Fit polynomial of degree 6
x_deg6, cov_deg6, apv_deg6, v_deg6 = fit_polynomial_curve(a, l, P, deg=6)
```

```python
# Step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 mark)

# Note: np.polyval expects coefficients in *descending* order (highest power first),
# but we built them in *ascending* order (a^0, a^1, ...), so we must reverse them.

quadratic_uw = np.polyval(x_deg2[::-1], a)
quartic_uw = np.polyval(x_deg4[::-1], a)
hexic_uw = np.polyval(x_deg6[::-1], a)
```

```python
# Step 8
# use matplotlib to plot each polynomial function, noisy observations l, true observations l_true, together in one single figure
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# Plot true observations
ax.plot(a, l_true, label="True observation f(a)", color='k', linewidth=2)

# Plot noisy observations
ax.scatter(a, l, label="Noisy observation f(a) + e", color='tab:blue', marker='o', alpha=0.3)

# Plot fitted polynomial curves
ax.plot(a, quadratic_uw, label="Quadratic fit (deg=2)", color='orange', linewidth=2)
ax.plot(a, quartic_uw, label="Quartic fit (deg=4)", color='green', linewidth=2)
ax.plot(a, hexic_uw, label="Hexic fit (deg=6)", color='red', linewidth=2)

# Add labels, legend, title
ax.set_xlabel("a (radians)")
ax.set_ylabel("Observation values")
ax.set_title("Polynomial Fits vs True and Noisy Observations")
ax.legend()

# Set x-axis limits
ax.set_xlim(-1, 1)

# Add grid and adjust layout
ax.grid(True, which='both', ls='-', color='0.65')
fig.tight_layout()

plt.show()
```

```python
# Step 9
# print the a posteriori variance factor for each polynomial function
# (3 marks)

print("A Posteriori Variance Factors (Goodness of Fit):")
print(f"Quadratic (deg=2): {apv_deg2:.6f}")
print(f"Quartic   (deg=4): {apv_deg4:.6f}")
print(f"Hexic     (deg=6): {apv_deg6:.6f}")
```

```python
# Step 10: Compute the norm (Euclidean distance) between predicted and true observations

# Calculate predicted values for each polynomial degree
y_pred_deg2 = np.polyval(x_deg2[::-1], a)
y_pred_deg4 = np.polyval(x_deg4[::-1], a)
y_pred_deg6 = np.polyval(x_deg6[::-1], a)


# Compute the norm for each degree
norm_deg2 = np.linalg.norm(y_pred_deg2 - l_true)
norm_deg4 = np.linalg.norm(y_pred_deg4 - l_true)
norm_deg6 = np.linalg.norm(y_pred_deg6 - l_true)


# Print the results
print("Norm between predicted and true observations:")
print(f"Quadratic (deg=2): {norm_deg2:.6f}")
print(f"Quartic   (deg=4): {norm_deg4:.6f}")
print(f"Hexic     (deg=6): {norm_deg6:.6f}")
```

```python
# Step 11: Comment on results

# Computed norms:
# Quadratic (deg=2): 229.075253
# Quartic   (deg=4): 38.398621
# Hexic     (deg=6): 17.777307
#
# The norm decreases as the polynomial degree increases from 2 to 6,
# showing that higher-degree models fit the data more accurately overall.
# However, the improvement between degree 4 and 6 is relatively small
# compared to the large improvement from degree 2 to 4.
#
# This suggests diminishing returns: increasing the model’s complexity
# beyond a certain point does not greatly reduce prediction error.
#
# In relation to the posterior variance factor, we can infer that the
# variance likely decreases with higher-degree polynomials, meaning the
# model becomes more confident in its predictions.
#
# Overall pattern:
# - Increasing degree → lower norm (better fit)
# - But beyond a point → smaller improvements and potential overfitting risk
```

```python
# step 14
# Compare the results of a posterier variance factor and norm from unweighted LSE and weighted LSE, what phenomena do you observe? Explain this phenomena.
# (2 marks)
#When we compare unweighted (ordinary LSE) and weighted LSE (weights = 1 / variance), two consistent effects are typically observed:
#Posteriori variance factor (apv):
#Weighted LSE usually produces a different apv than unweighted LSE. If observation variances vary a lot, weighting down the high-variance (noisy) measurements reduces their contribution to the residual sum used in apv, often lowering the apv relative to the unweighted solution — the fit is effectively judged by the more reliable measurements. Conversely, if weights are inappropriate, apv can increase.


#Norm between predicted and true observations:
#Weighted LSE prioritizes matching the reliable observations (small variance) and de-emphasizes noisy ones. Therefore: If the high-variance observations were dragging the unweighted fit away from the underlying truth, weighted LSE will typically reduce the norm to the true observations (better overall fidelity to true signal). If, however, the true signal in the high-variance observations differs systematically and carries useful information that the lower-variance points do not, weighted LSE can sometimes increase the overall norm to l_true.


#Observed phenomenon summary:
#In short, weighting by inverse variance reduces the influence of noisy measurements and usually yields a lower apv and improved fit to the truly reliable parts of the data. The improvement in the L2 norm tends to be greatest when measurement noise is uneven and dominated by a subset of high-variance observations; otherwise you may see only modest changes. This explains why, after applying P = diag(1/l_var), you should expect decreased posterior variance and a better (or at least more confident) fit to the reliable observations — but the exact change in the overall norm depends on which observations are most informative about the true signal.
```
